```python
import configparser
import os

from dotenv import load_dotenv
from huggingface_hub import InferenceClient
from jinja2 import Template
load_dotenv()
```

```python
hf_token = os.getenv("HF_TOKEN")
```

```python
SYSTEM_PROMPT = """
You are assisting in teaching genai fundamentals.
- Always be concise, simple and clear.
- If you don’t know, say so.
- Cite relevant context snippets if useful.
"""
```

```python
USER_PROMPT_TEMPLATE = """
User Question:
{{ query }}
"""

def render_user_prompt(query: str) -> str:
    template = Template(USER_PROMPT_TEMPLATE)
    return template.render(
        query=query
    )
```

```python
def load_config():
    try:
        # Create a ConfigParser object
        config = configparser.ConfigParser()
        # Read the config.ini file
        config_path = os.path.join("../../config/config.ini")
        config.read(config_path)
        return config
    except FileNotFoundError:
        raise
    except Exception:
        raise
```

```python
config = load_config()
llm = InferenceClient(
    model=config.get("LLM_MODEL", "model"),
    token=hf_token,
    timeout=config.getint("LLM_MODEL", "hub_timeout")
)
```

```python
output = llm.chat_completion(
    messages=[
        {"role": "system", "content": "You are a helpful assistant."},
        {"role": "user", "content": "Explain quantum computing in one sentence."}
    ],
    max_tokens=config.getint("LLM_MODEL", "max_tokens"),
    temperature=config.getfloat("LLM_MODEL", "temperature")
)

print(output.choices[0].message.content)
```
