## 2. Notebook documentation best practices — a before/after example

Compare these two versions of the same three lines of logic. Both *run*
identically; only one is actually usable by someone other than the person
who wrote it.


````python
# --- BEFORE: undocumented ---
def f(x, k=3):
    r = sorted(x, key=lambda d: d[1], reverse=True)
    return r[:k]

data = [("doc_a", 0.82), ("doc_b", 0.45), ("doc_c", 0.91)]
print(f(data))
````

The cell above runs fine, but a reader has to reverse-engineer what `f`
does, what `x` and `k` mean, and why `key=lambda d: d[1]` was chosen. Compare
to a documented version of the exact same logic:


````python
def top_k_by_score(scored_items: list, k: int = 3) -> list:
    """Return the k highest-scoring items from a list of (item, score)
    tuples, sorted descending by score.

    This is the same top-k selection pattern used throughout the Week 2/3
    retrieval notebooks -- e.g., picking the top-3 most relevant chunks
    after a semantic search.
    """
    return sorted(scored_items, key=lambda item_score: item_score[1], reverse=True)[:k]

retrieved_chunks_with_scores = [("doc_a", 0.82), ("doc_b", 0.45), ("doc_c", 0.91)]
print(top_k_by_score(retrieved_chunks_with_scores))
````

## 3. Structuring a GitHub repository

A reviewable repository typically looks like this:

```
my-genai-project/
├── README.md              <- what this is, how to run it, key results
├── requirements.txt        <- exact dependencies, so others can reproduce your environment
├── .gitignore              <- excludes secrets, __pycache__, large data files, etc.
├── LICENSE                 <- (optional but professional) an open-source license
├── notebooks/
│   ├── 01_rag_pipeline.ipynb
│   ├── 02_agent_workflow.ipynb
│   └── 03_evaluation.ipynb
├── src/                    <- reusable functions extracted out of notebooks, if any
│   └── retrieval_utils.py
```
